```python

%pip -q install pandas openpyxl scikit-learn

import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from google.colab import files

# ============================================================
# TRADITIONAL PROGRAMMING VS MACHINE LEARNING PROGRAMMING
# Group Members:
# 1. ANJALI CHOPDA - PRN: 1302250508
# 2. JASMEEN KAUR BATTH - PRN: 1302250678
# 3. ANKITA MAHAJAN - PRN: 1302250224
# 4. JANHAVI PATIL - PRN: 1302250163
# 5. AKHILANSH RACHAMALLA - PRN: 1302250384
# 6. HARSHIT BAHETI - PRN: 1302250365
# 7. HARDIK SHARMA - PRN: 1302250599
# ============================================================

# STEP 1: CREATE EXCEL INPUT DATA

data = pd.DataFrame({
    "Name": [
        "Student1", "Student2", "Student3", "Student4",
        "Student5", "Student6", "Student7", "Student8",
        "Student9", "Student10"
    ],
    "Maths": [80, 65, 30, 90, 35, 55, 25, 72, 45, 20],
    "Science": [75, 70, 45, 88, 30, 60, 35, 68, 42, 25],
    "English": [85, 60, 50, 92, 40, 58, 30, 75, 48, 35],
    "Result": [
        "PASS", "PASS", "FAIL", "PASS", "FAIL",
        "PASS", "FAIL", "PASS", "PASS", "FAIL"
    ]
})

excel_path = "/content/ml_input.xlsx"
data.to_excel(excel_path, index=False)
data = pd.read_excel(excel_path)

subjects = ["Maths", "Science", "English"]

print("\nSTEP 1: EXCEL INPUT DATA")
display(data)

# STEP 2: TRADITIONAL PROGRAMMING

def traditional_programming(maths, science, english):
    if maths >= 40 and science >= 40 and english >= 40:
        return "PASS"
    else:
        return "FAIL"

data["Traditional_Result"] = data.apply(
    lambda row: traditional_programming(
        row["Maths"], row["Science"], row["English"]
    ),
    axis=1
)

print("\nSTEP 2: TRADITIONAL PROGRAMMING RESULTS")
display(data[["Name", *subjects, "Traditional_Result"]])

# STEP 3: MACHINE LEARNING MODEL

X = data[subjects]
y = data["Result"].astype(str).str.strip().str.upper()

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

model = DecisionTreeClassifier(random_state=42, max_depth=3)
model.fit(X, y_encoded)

data["ML_Result"] = encoder.inverse_transform(model.predict(X))

print("\nSTEP 3: MACHINE LEARNING RESULTS")
display(data[["Name", *subjects, "Result", "ML_Result"]])

# STEP 4: COMPARISON

print("\nSTEP 4: TRADITIONAL VS MACHINE LEARNING")
display(data[
    ["Name", *subjects, "Result", "Traditional_Result", "ML_Result"]
])

# STEP 5: ENTER YOUR OWN STUDENT MARKS

print("\nSTEP 5: ENTER YOUR OWN STUDENT MARKS")
print("Enter 'stop' as the name to finish.")

while True:
    name = input("\nStudent name (or stop): ").strip()

    if name.lower() == "stop":
        break

    try:
        maths = float(input("Maths marks: "))
        science = float(input("Science marks: "))
        english = float(input("English marks: "))

        if not all(0 <= mark <= 100 for mark in [maths, science, english]):
            print("Marks must be between 0 and 100.")
            continue

        traditional_result = traditional_programming(
            maths, science, english
        )

        new_student = pd.DataFrame(
            [[maths, science, english]], columns=subjects
        )

        ml_result = encoder.inverse_transform(
            model.predict(new_student)
        )[0]

        print("\nPrediction for", name)
        print("Traditional Programming Result:", traditional_result)
        print("Machine Learning Prediction:", ml_result)

    except ValueError:
        print("Invalid input. Enter numeric marks.")

# STEP 6: ASK QUESTIONS ABOUT THE EXCEL DATA

print("\nSTEP 6: ASK QUESTIONS ABOUT THE DATA")
print("Examples:")
print("What is the average Maths mark?")
print("What is the average Science mark?")
print("What is the average English mark?")
print("How many students passed?")
print("How many students failed?")
print("Who scored highest in Maths?")
print("Who scored lowest in Science?")
print("Show students who passed.")
print("Show students who failed.")
print("Show all students.")
print("What is Student1's result?")
print("What is Student2's Maths mark?")
print("Type 'stop' to finish.")

while True:
    question = input("\nYour question: ").strip().lower()

    if question == "stop":
        print("Question section completed.")
        break

    if not question:
        print("Enter a question.")
        continue

    if "average" in question and "math" in question:
        print("Average Maths marks:", round(data["Maths"].mean(), 2))

    elif "average" in question and "science" in question:
        print("Average Science marks:", round(data["Science"].mean(), 2))

    elif "average" in question and "english" in question:
        print("Average English marks:", round(data["English"].mean(), 2))

    elif "how many" in question and "pass" in question:
        print("Students who passed:",
              int((data["Result"] == "PASS").sum()))

    elif "how many" in question and "fail" in question:
        print("Students who failed:",
              int((data["Result"] == "FAIL").sum()))

    elif "highest" in question and any(
        subject.lower() in question for subject in subjects
    ):
        subject = next(
            subject for subject in subjects
            if subject.lower() in question
        )
        row = data.loc[data[subject].idxmax()]
        print(row["Name"], "scored highest in", subject, ":", row[subject])

    elif "lowest" in question and any(
        subject.lower() in question for subject in subjects
    ):
        subject = next(
            subject for subject in subjects
            if subject.lower() in question
        )
        row = data.loc[data[subject].idxmin()]
        print(row["Name"], "scored lowest in", subject, ":", row[subject])

    elif "show" in question and "pass" in question:
        display(data[data["Result"] == "PASS"])

    elif "show" in question and "fail" in question:
        display(data[data["Result"] == "FAIL"])

    elif "show" in question and "all" in question:
        display(data)

    else:
        matched_students = [
            student for student in data["Name"]
            if student.lower() in question
        ]

        if matched_students:
            student_name = matched_students[0]
            row = data[data["Name"] == student_name].iloc[0]

            if "math" in question:
                print(student_name, "Maths marks:", row["Maths"])
            elif "science" in question:
                print(student_name, "Science marks:", row["Science"])
            elif "english" in question:
                print(student_name, "English marks:", row["English"])
            elif "result" in question or "pass" in question or "fail" in question:
                print("Actual Excel result:", row["Result"])
                print("Traditional result:", row["Traditional_Result"])
                print("ML prediction:", row["ML_Result"])
            else:
                display(row.to_frame().T)
        else:
            print("Question not recognized. Try an example above.")

# STEP 7: DOWNLOAD EXCEL FILE

print("\nSTEP 7: DOWNLOADING EXCEL FILE")
files.download(excel_path)

print("\nPROGRAM COMPLETED SUCCESSFULLY.")
```
