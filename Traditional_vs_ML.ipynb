```python

# ============================================================
# Traditional Programming vs Machine Learning Programming
# Group Members:
# 1. ANJALI CHOPDA - PRN: 1302250508
# 2. JASMEEN KAUR BATTH - PRN: 1302250678
# 3. ANKITA MAHAJAN - PRN: 1302250224
# 4. JANHAVI PATIL - PRN: 1302250163
# 5. AKHILANSH RACHAMALLA - PRN: 1302250384
# 6. HARSHIT BAHETI - PRN: 1302250365
# 7. HARDIK SHARMA - PRN: 1302250599
# ============================================================

%pip -q install pandas openpyxl scikit-learn

import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from google.colab import files

# 1. Create Excel input file (.xlsx)
data = pd.DataFrame({
    "Name": ["Student1", "Student2", "Student3", "Student4",
             "Student5", "Student6", "Student7", "Student8",
             "Student9", "Student10"],
    "Maths": [80, 65, 30, 90, 35, 55, 25, 72, 45, 20],
    "Science": [75, 70, 45, 88, 30, 60, 35, 68, 42, 25],
    "English": [85, 60, 50, 92, 40, 58, 30, 75, 48, 35],
    "Result": ["PASS", "PASS", "FAIL", "PASS", "FAIL",
               "PASS", "FAIL", "PASS", "PASS", "FAIL"]
})

data.to_excel("/content/ml_input.xlsx", index=False)

# 2. Read input from Excel
data = pd.read_excel("/content/ml_input.xlsx")
subjects = ["Maths", "Science", "English"]

print("INPUT DATA")
display(data)

# 3. Traditional Programming: predefined rules
def traditional_programming(row):
    if all(row[subject] >= 40 for subject in subjects):
        return "PASS"
    return "FAIL"

data["Traditional_Result"] = data.apply(
    traditional_programming, axis=1
)

# 4. Machine Learning Programming: Decision Tree
X = data[subjects]
y = data["Result"]

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y_encoded)

predictions = model.predict(X)
data["ML_Result"] = encoder.inverse_transform(predictions)

# 5. Display comparison
print("TRADITIONAL PROGRAMMING VS MACHINE LEARNING")
display(data[[
    "Name", "Maths", "Science", "English",
    "Result", "Traditional_Result", "ML_Result"
]])

print("Traditional programming uses predefined rules.")
print("Machine learning learns patterns from labeled data.")

# 6. Download the Excel input file
files.download("/content/ml_input.xlsx")
```
