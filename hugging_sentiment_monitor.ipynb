```python

# ============================================================
# Assignment: Hugging Face Sentiment Monitor
# Student Name: ANKITA MAHAJAN
# PRN: 1302250224
#
# Group Members:
# 1. ANJALI CHOPDA - PRN: 1302250508
# 2. JASMEEN KAUR BATTH - PRN: 1302250678
# 3. ANKITA MAHAJAN - PRN: 1302250224
# 4. JANHAVI PATIL - PRN: 1302250163
# 5. AKHILANSH RACHAMALLA - PRN: 1302250384
# 6. HARSHIT BAHETI - PRN: 1302250365
# 7. HARDIK SHARMA - PRN: 1302250599
# ============================================================

from transformers import pipeline


# ------------------------------------------------------------
# LOAD HUGGING FACE SENTIMENT MODEL
# ------------------------------------------------------------

print("Loading Hugging Face Sentiment Analysis Model...")

sentiment_analyzer = pipeline(
    "sentiment-analysis"
)


# ------------------------------------------------------------
# FUNCTION FOR SENTIMENT ANALYSIS
# ------------------------------------------------------------

def analyze_sentiment(text):

    result = sentiment_analyzer(text)[0]

    label = result["label"]
    score = result["score"]

    print("\nText:", text)
    print("Sentiment:", label)
    print("Confidence:", round(score, 4))


# ------------------------------------------------------------
# PROBLEM STATEMENT 1
# ------------------------------------------------------------

print("\n========== Problem Statement 1 ==========")

analyze_sentiment(
    "I really enjoyed using this application. It is very useful."
)


# ------------------------------------------------------------
# PROBLEM STATEMENT 2
# ------------------------------------------------------------

print("\n========== Problem Statement 2 ==========")

analyze_sentiment(
    "The service was very poor and I am disappointed."
)


# ------------------------------------------------------------
# PROBLEM STATEMENT 3
# ------------------------------------------------------------

print("\n========== Problem Statement 3 ==========")

analyze_sentiment(
    "The product is okay but could be improved."
)


# ------------------------------------------------------------
# PROBLEM STATEMENT 4
# ------------------------------------------------------------

print("\n========== Problem Statement 4 ==========")

analyze_sentiment(
    "The customer support team solved my problem quickly."
)


# ------------------------------------------------------------
# PROBLEM STATEMENT 5
# ------------------------------------------------------------

print("\n========== Problem Statement 5 ==========")

analyze_sentiment(
    "I am unhappy with the quality of the product."
)


# ------------------------------------------------------------
# PROBLEM STATEMENT 6
# ------------------------------------------------------------

print("\n========== Problem Statement 6 ==========")

analyze_sentiment(
    "The food was delicious and the staff were friendly."
)


# ------------------------------------------------------------
# PROBLEM STATEMENT 7
# ------------------------------------------------------------

print("\n========== Problem Statement 7 ==========")

analyze_sentiment(
    "The application crashes frequently and is frustrating to use."
)


# ------------------------------------------------------------
# PROBLEM STATEMENT 8
# ------------------------------------------------------------

print("\n========== Problem Statement 8 ==========")

analyze_sentiment(
    "The delivery arrived on time and the product was excellent."
)


# ------------------------------------------------------------
# ADDITIONAL PROBLEM STATEMENTS BY ANKITA MAHAJAN
# ------------------------------------------------------------

# PROBLEM STATEMENT 9: TRAVEL REVIEW ANALYSIS

print("\n========== Problem Statement 9 ==========")

analyze_sentiment(
    "The hotel was clean and the staff were very helpful."
)


# PROBLEM STATEMENT 10: ONLINE SHOPPING COMPLAINT

print("\n========== Problem Statement 10 ==========")

analyze_sentiment(
    "My order arrived damaged and I am very disappointed."
)


# PROBLEM STATEMENT 11: EDUCATIONAL FEEDBACK ANALYSIS

print("\n========== Problem Statement 11 ==========")

analyze_sentiment(
    "The teaching materials were clear and easy to understand."
)


# ------------------------------------------------------------
# USER INPUT
# ------------------------------------------------------------

print("\n========== Custom Sentiment Monitor ==========")

user_text = input(
    "\nEnter a sentence to analyze its sentiment: "
)

analyze_sentiment(user_text)
```
