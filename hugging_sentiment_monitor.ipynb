```python

# ============================================================
# ML FRAMEWORK: HUGGING FACE SENTIMENT MONITOR
# ============================================================
# Group Members:
# 1. ANJALI CHOPDA - PRN: 1302250508
# 2. JASMEEN KAUR BATTH - PRN: 1302250678
# 3. ANKITA MAHAJAN - PRN: 1302250224
# 4. JANHAVI PATIL - PRN: 1302250163
# 5. AKHILANSH RACHAMALLA - PRN: 1302250384
# 6. HARSHIT BAHETI - PRN: 1302250365
# 7. HARDIK SHARMA - PRN: 1302250599
# ============================================================

# STEP 1: Install the required library

%pip -q install transformers torch

# STEP 2: Import Hugging Face pipeline

from transformers import pipeline

# STEP 3: Load the pretrained sentiment-analysis model

print("Loading Hugging Face Sentiment Analysis Model...")

sentiment_monitor = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

print("Model loaded successfully!")

# STEP 4: Define the sentiment monitoring function

def analyze_sentiment(problem_statement, text):
    result = sentiment_monitor(text, truncation=True)[0]

    print("\nProblem Statement:", problem_statement)
    print("Input Text:", text)
    print("Sentiment:", result["label"])
    print("Confidence Score:", round(result["score"], 4))

# STEP 5: Problem Statement 1 - Product Review Analysis

analyze_sentiment(
    "Product Review Analysis",
    "This product is excellent and I am very happy with its quality."
)

# STEP 6: Problem Statement 2 - Customer Service Monitoring

analyze_sentiment(
    "Customer Service Monitoring",
    "The customer service was terrible and nobody helped me."
)

# STEP 7: Problem Statement 3 - Application Feedback Analysis

analyze_sentiment(
    "Application Feedback Analysis",
    "The application is easy to use and works perfectly."
)

# STEP 8: Problem Statement 4 - Social Media Comment Monitoring

analyze_sentiment(
    "Social Media Comment Monitoring",
    "I love this update. The new features are fantastic."
)

# STEP 9: Problem Statement 5 - Restaurant Feedback Analysis

analyze_sentiment(
    "Restaurant Feedback Analysis",
    "The food was cold and the service was very disappointing."
)

# STEP 10: Problem Statement 6 - Online Shopping Feedback

analyze_sentiment(
    "Online Shopping Feedback",
    "The delivery arrived on time and the product was excellent."
)

# STEP 11: Problem Statement 7 - Education Feedback Analysis

analyze_sentiment(
    "Education Feedback Analysis",
    "The lessons were clear and the teacher explained everything well."
)

# STEP 12: Problem Statement 8 - Complaint Monitoring

analyze_sentiment(
    "Complaint Monitoring",
    "I am extremely unhappy with this service and want a refund."
)

# STEP 13: Problem Statement 9 - Travel Review Analysis

analyze_sentiment(
    "Travel Review Analysis",
    "The hotel was comfortable and the staff were friendly."
)

# STEP 14: Problem Statement 10 - Healthcare Service Feedback

analyze_sentiment(
    "Healthcare Service Feedback",
    "The waiting time was too long and the service was disappointing."
)

# STEP 15: Analyze custom user input

print("\n========== CUSTOM SENTIMENT MONITOR ==========")

user_text = input("Enter a sentence to analyze: ")

if user_text.strip():
    analyze_sentiment("Custom User Input", user_text)
else:
    print("No text entered. Please run this cell again with a sentence.")
```
