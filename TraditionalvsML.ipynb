```python

# ============================================================
# Assignment: Traditional Programming vs Machine Learning
# Student Name: ANJALI CHOPDA
# PRN: 1302250508
#
# Group Members:
# 1. ANJALI CHOPDA - PRN: 1302250508
# 2. JASMEEN KAUR BATTH - PRN: 1302250678
# 3. ANKITA MAHAJAN - PRN: 1302250224
# 4. JANHAVI PATIL - PRN: 1302250163
# 5. AKHILANSH RACHAMALLA - PRN: 1302250384
# 6. HARSHIT BAHETI - PRN: 1302250365
# 7. HARDIK SHARMA - PRN: 1302250599
# ============================================================

import pandas as pd
from pathlib import Path
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# ------------------------------------------------------------
# 1. READ INPUT FROM EXCEL FILE
# ------------------------------------------------------------

file_name = "/content/ml_input.xlsx"

# Create the original Excel input file if it does not exist.
if not Path(file_name).exists():

    sample_data = [
        ["Student1", 80, 75, 85, "PASS"],
        ["Student2", 65, 70, 60, "PASS"],
        ["Student3", 30, 45, 50, "FAIL"],
        ["Student4", 90, 88, 92, "PASS"],
        ["Student5", 35, 30, 40, "FAIL"],
        ["Student6", 55, 60, 58, "PASS"],
        ["Student7", 25, 35, 30, "FAIL"],
        ["Student8", 72, 68, 75, "PASS"],
        ["Student9", 45, 42, 48, "PASS"],
        ["Student10", 20, 25, 35, "FAIL"]
    ]

    df = pd.DataFrame(
        sample_data,
        columns=[
            "Name", "Maths", "Science", "English", "Result"
        ]
    )

    df.to_excel(file_name, index=False)
    print("Original Excel input file created successfully.")

data = pd.read_excel(file_name)

print("========== INPUT DATA ==========")
print(data)
print()


# ------------------------------------------------------------
# 2. TRADITIONAL PROGRAMMING
# ------------------------------------------------------------
# Traditional programming uses predefined rules.
#
# Rule:
# If all subject marks are >= 40, result is PASS.
# Otherwise, result is FAIL.
#
# Improvement: Validate the input marks before predicting.
# ------------------------------------------------------------

def traditional_prediction(row):

    subjects = ["Maths", "Science", "English"]

    for subject in subjects:

        if subject not in row or pd.isna(row[subject]):
            raise ValueError(
                f"Missing marks for {subject}."
            )

        mark = float(row[subject])

        if not 0 <= mark <= 100:
            raise ValueError(
                f"{subject} marks must be between 0 and 100."
            )

    if all(float(row[subject]) >= 40 for subject in subjects):
        return "PASS"

    return "FAIL"


data["Traditional_Result"] = data.apply(
    traditional_prediction,
    axis=1
)


# ------------------------------------------------------------
# 3. MACHINE LEARNING PROGRAMMING
# ------------------------------------------------------------

features = ["Maths", "Science", "English"]

X = data[features]
y = data["Result"].astype(str).str.strip().str.upper()

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

model = DecisionTreeClassifier(random_state=42)

model.fit(X, y_encoded)

predictions = model.predict(X)

data["ML_Result"] = encoder.inverse_transform(predictions)


# ------------------------------------------------------------
# 4. DISPLAY RESULTS
# ------------------------------------------------------------

print("========== RESULTS ==========")

print(
    data[
        [
            "Name",
            "Maths",
            "Science",
            "English",
            "Traditional_Result",
            "ML_Result"
        ]
    ].to_string(index=False)
)

print()
print("Traditional Programming uses manually defined rules.")
print("Machine Learning learns patterns from the provided data.")

print("\n========== PREDICTION SUMMARY ==========")

print("\nTraditional Programming:")
print(data["Traditional_Result"].value_counts().to_string())

print("\nMachine Learning:")
print(data["ML_Result"].value_counts().to_string())

print("\nTotal students processed:", len(data))

print("\nProgram executed successfully.")
```
