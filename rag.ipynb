```python

# ============================================================
# Assignment: Retrieval Augmented Generation (RAG)
# Student Name: AKHILANSH RACHAMALLA
# PRN: 1302250384
#
# Group Members:
# 1. ANJALI CHOPDA - PRN: 1302250508
# 2. JASMEEN KAUR BATTH - PRN: 1302250678
# 3. ANKITA MAHAJAN - PRN: 1302250224
# 4. JANHAVI PATIL - PRN: 1302250163
# 5. AKHILANSH RACHAMALLA - PRN: 1302250384
# 6. HARSHIT BAHETI - PRN: 1302250365
# 7. HARDIK SHARMA - PRN: 1302250599
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ============================================================
# KNOWLEDGE BASE
# ============================================================

documents = [
    {
        "title": "Machine Learning",
        "text": (
            "Machine Learning is a branch of Artificial Intelligence. "
            "It allows computers to learn patterns from data."
        )
    },
    {
        "title": "Supervised Learning",
        "text": (
            "Supervised learning uses labelled training data. "
            "Classification and regression are common supervised "
            "learning tasks."
        )
    },
    {
        "title": "Unsupervised Learning",
        "text": (
            "Unsupervised learning works with data that does not have "
            "labelled outputs. Clustering is a common example."
        )
    },
    {
        "title": "Natural Language Processing",
        "text": (
            "Natural Language Processing allows computers to process "
            "and understand human language."
        )
    },
    {
        "title": "Retrieval Augmented Generation",
        "text": (
            "Retrieval Augmented Generation retrieves relevant "
            "information from a knowledge base before providing "
            "an answer."
        )
    },

    # Additional topics added by Akhilansh.

    {
        "title": "Model Evaluation",
        "text": (
            "Model evaluation measures how well a machine learning "
            "model performs using suitable metrics and data that "
            "were not used for training."
        )
    },
    {
        "title": "Data Preprocessing",
        "text": (
            "Data preprocessing prepares raw data for analysis by "
            "handling missing values, correcting formats, and "
            "transforming features."
        )
    },
    {
        "title": "Overfitting",
        "text": (
            "Overfitting occurs when a model learns training data "
            "too closely and performs poorly on new, unseen data."
        )
    }
]


# ============================================================
# CREATE TF-IDF VECTOR REPRESENTATION
# ============================================================

vectorizer = TfidfVectorizer(stop_words="english")

document_vectors = vectorizer.fit_transform(
    [document["text"] for document in documents]
)


# ============================================================
# RETRIEVAL FUNCTION
# ============================================================

def retrieve_information(question):

    question_vector = vectorizer.transform([question])

    similarity_scores = cosine_similarity(
        question_vector,
        document_vectors
    )[0]

    best_index = similarity_scores.argmax()
    best_score = similarity_scores[best_index]

    return documents[best_index], best_score


# ============================================================
# RAG RULES
# ============================================================

def apply_rag_rules(question):

    # RULE 1: Reject empty questions.
    if not question.strip():
        return "Please enter a valid question."

    # RULE 2: Retrieve the most relevant information.
    retrieved_document, score = retrieve_information(question)

    # RULE 3: Reject information below the relevance threshold.
    if score < 0.15:
        return (
            "Information not found in the provided knowledge base."
        )

    # RULE 4: Answer using retrieved information.
    answer = retrieved_document["text"]

    # RULE 5: Identify the source of the answer.
    source = retrieved_document["title"]

    # RULE 6: Display the similarity score.
    return (
        f"Answer based on retrieved information:\n{answer}\n"
        f"Source: {source}\n"
        f"Similarity Score: {score:.3f}"
    )


# ============================================================
# DISPLAY RAG INFORMATION
# ============================================================

print("=" * 55)
print("RETRIEVAL AUGMENTED GENERATION (RAG)")
print("Student: AKHILANSH RACHAMALLA")
print("PRN: 1302250384")
print("=" * 55)

print("\nKnowledge Base Topics:")

for document in documents:
    print("-", document["title"])

print("\nRules applied:")
print("1. Empty questions are rejected.")
print("2. Relevant information is retrieved from the knowledge base.")
print("3. Low-relevance results are rejected.")
print("4. Answers use retrieved information.")
print("5. The source title is displayed.")
print("6. The similarity score is displayed.")
print("7. Outside information is not retrieved.")
print("8. The knowledge base determines the answer.")
print("9. Missing information is reported.")
print("10. Retrieved text is shown without inventing new facts.")


# ============================================================
# USER QUESTIONS
# ============================================================

while True:

    question = input(
        "\nAsk a question (type 'exit' to stop): "
    )

    if question.strip().lower() == "exit":
        print("RAG system stopped.")
        break

    result = apply_rag_rules(question)

    print("\n" + result)
```
