```python

# ============================================================
# ML FRAMEWORK: RETRIEVAL AUGMENTED GENERATION (RAG)
# ============================================================
# Group Members:
# 1. ANJALI CHOPDA - PRN: 1302250508
# 2. JASMEEN KAUR BATTH - PRN: 1302250678
# 3. ANKITA MAHAJAN - PRN: 1302250224
# 4. JANHAVI PATIL - PRN: 1302250163
# 5. AKHILANSH RACHAMALLA - PRN: 1302250384
# 6. HARSHIT BAHETI - PRN: 1302250365
# 7. HARDIK SHARMA - PRN: 1302250599
# ============================================================

# STEP 1: Install required library

%pip -q install scikit-learn

# STEP 2: Import libraries

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# STEP 3: Create the knowledge base

documents = [
    {
        "title": "Machine Learning",
        "text": "Machine Learning is a branch of Artificial Intelligence that enables computers to learn patterns from data."
    },
    {
        "title": "Supervised Learning",
        "text": "Supervised learning uses labelled training data. Classification and regression are common supervised learning tasks."
    },
    {
        "title": "Unsupervised Learning",
        "text": "Unsupervised learning uses data without labelled outputs. Clustering is a common example."
    },
    {
        "title": "Natural Language Processing",
        "text": "Natural Language Processing enables computers to process and understand human language."
    },
    {
        "title": "Retrieval Augmented Generation",
        "text": "Retrieval Augmented Generation retrieves relevant information from a knowledge base and uses that context to support an answer."
    },
    {
        "title": "Decision Tree",
        "text": "A Decision Tree is a supervised machine learning algorithm that makes predictions using rules based on input features."
    },
    {
        "title": "Sentiment Analysis",
        "text": "Sentiment analysis identifies opinions expressed in text, such as positive and negative sentiment."
    }
]

# STEP 4: Convert documents into numerical vectors

vectorizer = TfidfVectorizer(stop_words="english")
document_vectors = vectorizer.fit_transform(
    [document["text"] for document in documents]
)

# STEP 5: Retrieve relevant information

def retrieve_information(question):
    question_vector = vectorizer.transform([question])

    similarity_scores = cosine_similarity(
        question_vector, document_vectors
    )[0]

    best_index = similarity_scores.argmax()
    best_score = float(similarity_scores[best_index])

    return documents[best_index], best_score

# STEP 6: RAG function with additional rules

def rag_answer(question):

    # Rule 1: Reject empty questions.
    if not question.strip():
        return "Please enter a valid question."

    # Rule 2: Limit question length.
    if len(question) > 1000:
        return "Question must not exceed 1000 characters."

    # Rule 3: Retrieve information from the knowledge base.
    document, score = retrieve_information(question)

    # Rule 4: Reject low-relevance results.
    if score < 0.15:
        return "Information not found in the provided knowledge base."

    # Rule 5: Use the retrieved document as the answer.
    answer = document["text"]

    # Rule 6: Display the source document.
    source = document["title"]

    # Rule 7: Display the similarity score.
    similarity = round(score, 3)

    # Rule 8: Do not add unsupported information.
    # The answer is taken directly from the retrieved document.

    # Rule 9: Return the answer in a clear format.
    return (
        f"Answer: {answer}\n"
        f"Source: {source}\n"
        f"Similarity Score: {similarity}"
    )

# STEP 7: Display the additional rules

print("========== RAG SYSTEM ==========")
print("Rule 1: Reject empty questions.")
print("Rule 2: Limit question length to 1000 characters.")
print("Rule 3: Retrieve information from the knowledge base.")
print("Rule 4: Reject low-relevance results.")
print("Rule 5: Use retrieved information as the answer.")
print("Rule 6: Display the source document.")
print("Rule 7: Display the similarity score.")
print("Rule 8: Do not add unsupported information.")
print("Rule 9: Return answers in a clear format.")
print("Rule 10: Use the same retrieval process for repeated questions.")

# STEP 8: Display example questions

print("\n========== EXAMPLE QUESTIONS ==========")
print("1. What is Machine Learning?")
print("2. What is Supervised Learning?")
print("3. What is Unsupervised Learning?")
print("4. What is Natural Language Processing?")
print("5. What is Retrieval Augmented Generation?")
print("6. What is a Decision Tree?")
print("7. What is Sentiment Analysis?")

print("\nEnter an example question or type your own.")
print("Type 'exit' to stop the program.")

# STEP 9: Interactive question-and-answer system

while True:
    question = input("\nEnter your question: ")

    if question.strip().lower() == "exit":
        print("RAG system stopped. Thank you!")
        break

    print("\n========== RAG ANSWER ==========")
    print(rag_answer(question))
```
